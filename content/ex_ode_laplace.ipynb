```python
from sympy import *
from sympy.integrals.laplace import laplace_correspondence, laplace_initial_conds
from display_utils import mprint

t, s = symbols('t s', positive=True)
y = Function('y')
Y = Function('Y')
 
# Step 1: the ODE (implicitly = 0)
#   y'' - y' - 6y = 0,   y(0) = -1,   y'(0) = 0
ode = y(t).diff(t, 2) - y(t).diff(t) - 6*y(t)
 
# Step 2: Laplace transform the ODE
F = laplace_transform(ode, t, s, noconds=True)
F = laplace_correspondence(F, {y: Y})           # y(t) -> Y(s)
F = laplace_initial_conds(F, t, {y: [-1, 0]})   # [y(0), y'(0)]
print("Laplace-transformed ODE = 0:")
mprint(F)
 
# Step 3: solve for Y(s)
Ys = apart(solve(F, Y(s))[0], s)
print("\nY(s):")
mprint(Ys)
 
# Step 4: inverse Laplace transform -> y(t)
yt = simplify(inverse_laplace_transform(Ys, s, t))
print("\ny(t):")
mprint(yt)
```
