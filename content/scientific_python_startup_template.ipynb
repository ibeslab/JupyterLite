## Starting point: plot a NumPy array with Matplotlib

The most basic building block -- generate some data, plot it, label it.


```
x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, y, label='sin(x)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Quick NumPy + Matplotlib check')
ax.legend()
ax.grid(True)
plt.show()
```

## Multiple series on one plot

Common next step -- comparing a couple of signals.


```
y2 = np.cos(x)

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, y, label='sin(x)')
ax.plot(x, y2, label='cos(x)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
ax.grid(True)
plt.show()
```
